```python
import re
import nltk
import matplotlib.pyplot as plt
from collections import Counter
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')

```

```python
text = """
Natural Language Processing is a part of Artificial Intelligence.
NLP is used in machine learning, chatbots, speech recognition,
text classification, and many real world applications.
Machine learning models use data for prediction.
"""
```

```python
from google.colab import files
uploaded = files.upload()

file_name = list(uploaded.keys())[0]

with open(file_name, 'r', encoding='utf-8') as file:
    text = file.read()
```

```python
text = text.lower()
```

```python
text = re.sub(r'[^a-z\s]', '', text)

```

```python
tokens = text.split()
print("Tokens:\n", tokens)
```

```python
stop_words = set(stopwords.words('english'))

filtered_tokens = [word for word in tokens if word not in stop_words]

print("\nAfter Stopword Removal:\n", filtered_tokens)
```

```python
word_freq = Counter(filtered_tokens)
```

```python
top_50 = word_freq.most_common(50)

print("\nTop 50 Frequent Words:\n")
for word, freq in top_50:
    print(word, ":", freq)
```

```python
words = [word for word, freq in top_50]
frequencies = [freq for word, freq in top_50]

plt.figure(figsize=(15,6))
plt.bar(words, frequencies)
plt.xticks(rotation=90)
plt.title("Top 50 Frequent Words")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.show()
```
