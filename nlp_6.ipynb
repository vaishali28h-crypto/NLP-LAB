```
import spacy
nlp = spacy.load("en_core_web_sm")
```

```
text = """
Elon Musk founded Tesla in 2003 in the United States.
Natural Language Processing helps computers understand human language.
"""

doc = nlp(text)
```

```
print(" TOKENIZATION:\n")

for token in doc:
    print(token.text)
```

```
print("\n STOP WORD REMOVAL:\n")

filtered_tokens = [token.text for token in doc
                   if not token.is_stop and not token.is_punct]

print(filtered_tokens)
```

```
print("\n LEMMATIZATION:\n")

lemmatized_tokens = [token.lemma_ for token in doc
                     if not token.is_stop and not token.is_punct]

print(lemmatized_tokens)
```

```
print("\n POS TAGGING:\n")

for token in doc:
    print(f"{token.text:15} {token.pos_:10}")
```

```
print("\n NAMED ENTITY RECOGNITION (NER):\n")

for ent in doc.ents:
    print(f"{ent.text:20} {ent.label_}")
```
